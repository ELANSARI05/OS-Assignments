```python
import dash
import pandas
from dash import dcc, html, Input, Output, State, ctx
import plotly.express as px
import pandas as pd
from code import Process, fcfs, generate_random_processes  # use your module's code

app = dash.Dash(__name__)
app.title = "CPU Scheduling Simulator"

app.layout = html.Div([
    html.H1("CPU Scheduler"),

    # Input mode
    html.Label("Choose input method:"),
    dcc.RadioItems(
        id='input_mode',
        options=[
            {'label': 'Manual', 'value': 'manual'},
            {'label': 'Random', 'value': 'random'},
            {'label': 'From File', 'value': 'file'}
        ],
        value='manual',
        inline=True
    ),
    html.Div(id='input_area'),

    html.Br(),
    html.Label("Choose Algorithm:"),
    dcc.Dropdown(
        id='algorithm',
        options=[
            {'label': 'FCFS', 'value': 'fcfs'},
            {'label': 'SJF', 'value': 'sjf'},
            {'label': 'Priority', 'value': 'priority'},
            {'label': 'Round Robin', 'value': 'rr'},
            {'label': 'Priority + RR', 'value': 'priority_rr'}
        ],
        value='fcfs'
    ),
    dcc.Input(id='quantum', type='number', placeholder='Quantum (if RR)', debounce=True),

    html.Button("Run Scheduler", id='run_btn'),

    html.H3("Gantt Chart"),
    dcc.Graph(id='gantt_chart'),

    html.H3("Performance Metrics"),
    html.Div(id='metrics')
])
@app.callback(
    Output('input_area', 'children'),
    Input('input_mode', 'value')
)
def display_input_form(mode):
    if mode == 'manual':
        return html.Div([
            html.Label("Number of processes:"),
            dcc.Input(id='num_manual', type='number', value=3),
            html.Button("Generate Form", id='gen_form')
        ])
    elif mode == 'random':
        return html.Div([
            html.Label("Random: How many processes?"),
            dcc.Input(id='num_random', type='number', value=5)
        ])
    elif mode == 'file':
        return dcc.Upload(id='upload', children=html.Button("Upload File"))

def create_gantt(processes):
    df = pd.DataFrame([{
        "Task": f"P{p.pid}",
        "Start": p.start_time,
        "Finish": p.completion_time
    } for p in processes])
    fig = px.timeline(df, x_start="Start", x_end="Finish", y="Task", color="Task")
    fig.update_layout(xaxis_title="Time", yaxis_title="Processes", showlegend=False)
    return fig



# More callbacks:
# - Generate manual form
# - Read and store process list
# - Run selected algorithm
# - Generate Gantt chart
# - Compute performance metrics
```
